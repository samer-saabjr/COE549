````python
from pydantic import BaseModel, Field
# Field is a function in the Pydantic library used to customize and add metadata to fields within a Pydantic model. While Pydantic automatically infers type information from Python type hints, Field allows for more granular control over individual fields.
from langchain_core.messages import HumanMessage, SystemMessage

from llms_course.utils.llms import get_structured_output_llm
````

Both the Pydantic model's docstring, each Field's description, and the names of the variables will be provided to the LLM. You can use them to further guide its behavior. 

It's recommended to pick clear and descriptive variable names, even if they're a bit long.

````python
class Joke(BaseModel):
    """
    A joke with a setup and a punchline.
    """
    setup: str = Field(description = "The setup of the joke.")
    punchline: str = Field(description = "The punchline of the joke.")
    # If you use an Optional here, the LLM can decide to not include the punchline.
````

````python
def generate_joke(topic: str) -> Joke:
    """
    Generate a joke with a setup and a punchline based on the given topic using an LLM.

    Args:
        topic (str): The topic of the joke.

    Returns:
        (Joke): The joke object containing a setup and a punchline.
    """
    llm = get_structured_output_llm(Joke)

    system_prompt = """# Role
You are a helpful assistant that generates jokes.

# Instructions
- Generate a joke based on the given topic.
"""

    messages = [
        SystemMessage(content = system_prompt),
        HumanMessage(content = f"Generate a joke based on the given topic: ```{topic}```")
    ]

    response: Joke = llm.invoke(messages)
    setup = response.setup
    punchline = response.punchline
    joke = f"{setup}\n{punchline}"
    return joke
````

````python
joke = generate_joke("LLMs")
print(joke)
````

We can guide the LLM's behavior using the Pydantic's values directly. 

The LLM is more likely to adhere to instructions placed here than the main system prompt.

````python
class KnockKnockJoke(BaseModel):
    """
    A knock-knock joke.

    Example:
    Knock knock.
    Who's there?
    Interrupting cow. # `whos_there` field
    Interrupting cow who? # `who` field
    MOO! # `punchline` field
    """
    whos_there: str = Field(description = "The who's there part of the knock-knock joke.")
    who: str = Field(description = "The who part of the knock-knock joke.")
    punchline: str = Field(description = "The punchline of the knock-knock joke.")
````

Note that we use the exact same generic system prompt here as the previous example.

````python
def generate_knock_knock_joke(topic: str) -> KnockKnockJoke:
    """
    Generate a knock-knock joke based on the given topic using an LLM.

    Args:
        topic (str): The topic of the joke.

    Returns:
        (KnockKnockJoke): The knock-knock joke object containing a who's there part, a who part, and a punchline.
    """
    llm = get_structured_output_llm(KnockKnockJoke)

    system_prompt = """# Role
You are a helpful assistant that generates jokes.

# Instructions
- Generate a joke based on the given topic.
"""

    messages = [
        SystemMessage(content = system_prompt),
        HumanMessage(content = f"Generate a joke based on the given topic: ```{topic}```")
    ]

    response: KnockKnockJoke = llm.invoke(messages)
    whos_there = response.whos_there
    who = response.who
    punchline = response.punchline
    knock_knock_joke = f"Knock knock.\nWho's there?\n{whos_there}\n{who}\n{punchline}"
    return knock_knock_joke
````

````python
knock_knock_joke = generate_knock_knock_joke("Artificial Intelligence")
print(knock_knock_joke)
````
