```python
from pydantic import BaseModel 
# BaseModel is a fundamental class in the Pydantic library used for defining data models and enforcing data validation in Python.
# It serves as the base class for creating Pydantic models, which are essentially Python classes that use type annotations to specify the expected structure and types of data.

from langgraph.graph import StateGraph, END
# StateGraph manages a central, shared state object that is accessible to all nodes within the graph. 
# This state acts as a persistent memory, allowing information to be stored and updated across different stages of the workflow.
# END is a special node that represents the end of the graph
```

```python
# Instead of making a normal Python class and writing your own checks:

# class State:
#     def __init__(self, message):
#         self.message = message

# Pydantic gives you BaseModel

class State(BaseModel):
    message: str
```

```python
def reverse_message(state: State) -> State:
    """
    Reverse the message.

    Args:
        state (State): The current state of the graph.

    Returns:
        (State): The updated state of the graph.
    """
    reversed_message = state.message[::-1]
    state.message = reversed_message
    return state
```

```python
# Create the graph builder
graph_builder = StateGraph(State)

# Add the `reverse_message` node
graph_builder.add_node("reverse_message", reverse_message)

# Add an edge from the `reverse_message` node to END
graph_builder.add_edge("reverse_message", END)

# Set the entry point to the `reverse_message` node
graph_builder.set_entry_point("reverse_message")

# Compile the graph
graph = graph_builder.compile()
```

```python
# Input into the graph must match the state schema
inputs = {"message": "Hello, world!!"}

# Invoke the graph with the inputs and return the final state
final_state = graph.invoke(inputs)

print(final_state) # Returns a dictionary with the final state. Should print {"message": "!dlrow ,olleH"}
```
